```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(30,9))
sns.histplot(data=df,x="Age",bins=10,color="green",kde=True)
plt.title("hist")
plt.xlabel("Age")
plt.ylabel("count")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
Pclass_count=df["Pclass"].value_counts()
plt.figure(figsize=(30,9))
Pclass_count.plot(kind="bar",color=["green","red"])

plt.title("hist")
plt.xlabel("Age")
plt.ylabel("count")
plt.show()
```

```python
plt.figure(figsize=(30,9))
sns.scatterplot(data=df,x="Age",y="Name",color="green")
plt.title("hist")
plt.xlabel("Age")
plt.ylabel("count")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
df_sorted=df.sort_values("Pclass")
plt.figure(figsize=(30,9))
plt.plot(df_sorted["Pclass"],color="green",linewidth=4)

plt.title("hist")
plt.xlabel("Age")
plt.ylabel("count")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,8))
sns.boxplot(x="Age",color="green",data=df)
plt.title("hist")
plt.xlabel("Age")
plt.ylabel("count")
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.violinplot(x="Age",color="green",data=df)
plt.title("hist")
plt.xlabel("Age")
plt.ylabel("count")
plt.show()
```
